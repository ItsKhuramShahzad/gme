```python
from scipy.optimize import minimize
```

```python
def loss(x):
    return (x[0]-3)**2 + (x[1]+2)**2

result = minimize(loss, x0= [0,0],
                 method = "L-BFGS-B", 
                 bounds= [(-10, 10), (-5, 5)])
print(result.x)
print(result.fun)
```
